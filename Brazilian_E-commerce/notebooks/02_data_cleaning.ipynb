```python
import pandas as pd

orders = pd.read_csv("C:/Users/HLOGIZNBUCKS/Downloads/Data-Analytics-Training/Brazilian_E-commerce/olist_orders_dataset.csv")
order_items = pd.read_csv("C:/Users/HLOGIZNBUCKS/Downloads/Data-Analytics-Training/Brazilian_E-commerce/olist_order_items_dataset.csv")
products = pd.read_csv("C:/Users/HLOGIZNBUCKS/Downloads/Data-Analytics-Training/Brazilian_E-commerce/olist_products_dataset.csv")
customers = pd.read_csv("C:/Users/HLOGIZNBUCKS/Downloads/Data-Analytics-Training/Brazilian_E-commerce/olist_customers_dataset.csv")

# Lets merge all tables
df = order_items.merge(orders, on='order_id') \
                .merge(products, on='product_id') \
                .merge(customers, on='customer_id')

# Lets check for missing values
df.isnull().sum()

# Lets fix data types
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])

# Lets create Business metrics

df['revenue'] = df['price'] + df['freight_value']
print(df['revenue'])

df['cost'] = df['freight_value']
print(df['cost'])

df['profit'] = df['price'] - df['freight_value']
print(df['profit'])

df['profit_margin'] = df['profit'] / df['revenue']
print(df['profit_margin'])

df['delivery_time'] = (
    df['order_delivered_customer_date'] - df['order_purchase_timestamp']
).dt.days

df['order_month'] = df['order_purchase_timestamp'].dt.to_period('M')

# Lets remove rows with missing delivery dates
df = df[df['order_delivered_customer_date'].notnull()]

# Lets remove negative or zero revenue
df = df[df['revenue'] > 0]

df[['revenue','profit','delivery_time']].describe()
df.to_csv("C:/Users/HLOGIZNBUCKS/Downloads/Data-Analytics-Training/Brazilian_E-commerce/data/cleaned_data.csv", index=False)
```
